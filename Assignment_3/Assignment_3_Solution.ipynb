# 1. Inventory Price Calculator

```python
""" 
Task:
Write a function calculate_inventory_summary (prices) that:
 receives a list of product prices
 returns total_cost and average_cost

"""

def calculate_inventory_summary(prices):
  
    total_cost = sum(prices)
    
  
    if len(prices) > 0:
        average_cost = total_cost / len(prices)
    else:
        average_cost = 0
    
    return total_cost, average_cost

prices = [120, 250, 399, 150]
result = calculate_inventory_summary(prices)
print(result)
```

# 2. Unique Visitor Tracker

```python
"""  
Task:
Write a function count_unique_visitors(visitor_list) that:
 takes a list of visitor IDs
 returns the number of unique visitors

"""

def count_unique_visitors(visitor_list):
  unique_visitors = set(visitor_list)
  return len(unique_visitors)

input_list = [101, 205, 101, 310, 205, 550, 420, 101]
unique_count = count_unique_visitors(input_list)

print(f"Input List: {input_list}")
print(f"Number of Unique Visitors: {unique_count}")
```

# 3. Product Category Counter

```python
"""   
Task:
Write a function count_categories(products) that:
 takes a list of product names
 returns a dictionary counting each product type.

"""


def count_categories(products):
    category_count = {}
    
    for item in products:
        if item in category_count:
            category_count[item] += 1
        else:
            category_count[item] = 1
            
    return category_count


products = ["Laptop", "Smartphone", "Monitor", "Laptop", "Smartphone", "Headphones", "Laptop"]
result = count_categories(products)
print(result)



```

# 4. Temperature Analytics

```python
"""    
Task:
Write a function analyze_temperatures(temp_list) that:
 separates high (> 38°C) and normal temps
 returns two lists: (high_list, normal_list)

"""

def analyze_temperatures(temp_list):
    high_list = []
    normal_list = []
    
    for temp in temp_list:
        if temp > 38:
            high_list.append(temp)
        else:
            normal_list.append(temp)
            
    return high_list, normal_list



temps = [36.5, 38.0, 39.1, 37.2, 40.5, 38.0, 37.9]
result = analyze_temperatures(temps)
print(result)
```

# 5. Menu Price Lookup

```python
"""   
Task:
Write a function get_price(menu, item) that:
 receives a dictionary menu and an item name
 returns the price if exists
 otherwise returns "Item not found"

"""

def get_price(menu, item):
   
    return menu.get(item, "Item not found")


menu1 = {"Coffee": 3.50, "Tea": 2.75, "Muffin": 4.00, "Sandwich": 8.50}
item1 = "Muffin"
price1 = get_price(menu1, item1)

print("--- Example 1: Item Found ---")
print(f"Menu: {menu1}")
print(f"Lookup Item: '{item1}'")
print(f"Output: {price1}")
print("-" * 30)


menu2 = {"Coffee": 3.50, "Tea": 2.75, "Muffin": 4.00, "Sandwich": 8.50}
item2 = "Donut"
price2 = get_price(menu2, item2)

print("--- Example 2: Item Not Found ---")
print(f"Menu: {menu2}")
print(f"Lookup Item: '{item2}'")
print(f"Output: {price2}")
print("-" * 30)
```

# 6. Student Grading System

```python
"""   
Task:
Write a function calculate_grade(score) that:
 returns "A+", "A", "A-", "F" based on numeric marks
Write another function grade_students(student_dict) to:
 take a dictionary of student:score
 return student:grade dictionary

"""

def calculate_grade(score):
    if score >= 80:
        return "A+"
    elif score >= 70:
        return "A"
    elif score >= 60:
        return "A-"
    else:
        return "F"


def grade_students(student_dict):
    graded = {}
    
    for name, score in student_dict.items():
        graded[name] = calculate_grade(score)
    
    return graded

students = {"Alice": 85, "Bob": 78, "Charlie": 32, "David": 69}
result = grade_students(students)
print(result)
```

# 7. Discount Calculator

```python
"""  
Task:
Write a function apply_discounts(products) where:
 products is a list of tuples: (name, price, discount)
 return a list of updated prices after discount  

"""
def apply_discounts(products):
    final_prices = []
    
    for name, price, discount in products:
        discounted_price = price - (price * discount / 100)
        final_prices.append(round(discounted_price, 2))
    
    return final_prices


items = [
    ("Keyboard", 80.00, 20),
    ("Mouse", 25.00, 10),
    ("Monitor", 300.00, 5)
]

result = apply_discounts(items)
print("List of prices: ", result)  
```

# 8. Hashtag Generator

```python
"""    
Task:
Write a function generate_hashtags(sentence) that:
 converts every word into a hashtag
 returns a list of hashtags

"""

def generate_hashtags(sentence):
    words = sentence.split()
    hashtags = []
    
    for word in words:
        hashtags.append("#" + word.lower())
    
    return hashtags

text = "AI Data Science"
result = generate_hashtags(text)
print(result) 
```

# 9. User Authentication Mock

```python
"""
Task:
Write a function login(username, password) that:
 checks if username exists
 checks if password matches
 returns "Login Successful" or "Invalid Credentials"     

"""

CREDENTIALS = {
    "alex_s": "securepwd1",
    "maria_t": "password101",
    "john_d": "johndoe123"
}

def login(username, password):

    if username in CREDENTIALS:

        if CREDENTIALS[username] == password:
            return "Login Successful"
        else:
            return "Invalid Credentials"
    else:
        return "Invalid Credentials"


print(login("maria_t", "password101"))
print(login("alex_s", " securepwd12"))
print(login("peter_z", "any_password"))
```

# 10. Shopping Cart Bill Calculator

```python
"""    
Task:
Write a function calculate_total(cart, prices) that:
 takes a shopping cart list
 takes a dictionary of item:price
 returns the total bill amount

"""
def calculate_total(cart, prices):
    total = 0
    for item in cart:
        if item in prices:
            total += prices[item]
        else:
            print(f"Warning: '{item}' not found in price list.")
    return total



cart = ["Milk", "Bread", "Milk", "Cheese", "Bread"]
prices = {
    "Milk": 3.00,
    "Bread": 2.50,
    "Eggs": 4.00,
    "Cheese": 5.00
}

total_bill = calculate_total(cart, prices)


print("Total Bill Amount:", total_bill)

```
